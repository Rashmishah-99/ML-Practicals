```python
def factorial(num):
  if num==1:
    return 1
  else:
    return (num * factorial(num-1))
x=int(input('enter a number:'))
print('factorial of',x,'is',factorial(x))
```

# **4) Implement a function that checks if a given string is pangram (contains all letters of the alphabet).**

```python
def pangram(str):
  alphabet="abcdefghijklmnopqrstuvwxyz"
  for char in alphabet:
    if char not in str.lower():
      return False
    return True
string='The quick brown fox jumps over the lazy dog'
if(pangram(string))==True:
  print('The string is pangram')
else:
  print('The string is not pangram.')
```

# **5) Given a string, write a function to remove all vowels from it and return the modified string.**

```python
def vowels(str):
  vowels=['a','e','i','o','u','A','E','I','O','U']
  new_str=''
  for char in str:
    if char not in vowels:
      new_str=new_str+char
  return new_str
string='Computer Science Engineering'
print('Modified string:',vowels(string))
```
